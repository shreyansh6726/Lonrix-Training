In deep learning, an **optimizer** is an algorithm used to change the attributes of a neural network—specifically its **weights and biases**—to minimize the loss function. Think of the optimizer as the navigation system guiding the model down the error mountain to find the lowest possible valley (global minimum).

---

### Understanding the Key Ones You Mentioned

1. **SGD (Stochastic Gradient Descent):** The foundational optimizer. Instead of looking at the entire dataset to compute a gradient (which is computationally impossible for large data), SGD looks at a single random sample or a small batch, calculates the error gradient, and takes a step downhill. It is simple, but can sometimes get stuck in local minima or zigzag wildly in steep ravines.
2. **RMSprop (Root Mean Square Propagation):** An adaptive learning rate optimizer. It solves SGD's zigzagging problem by dividing the learning rate by a running average of the magnitudes of recent gradients. This allows the algorithm to move faster in gently sloping directions and slow down in steep ones.
3. **Adam (Adaptive Moment Estimation):** Currently the most popular default optimizer in deep learning. It combines the best ideas of **Momentum** (remembering past directions to maintain speed) and **RMSprop** (scaling the learning rate based on recent gradient magnitudes). It tracks both the first moment (the mean) and the second moment (the uncentered variance) of the gradients.

---

### The Complete Map of Neural Network Optimizers

While research papers introduce dozens of variants, optimizers generally fall into a few clear evolutionary categories:

| Category | Optimizer Name | Key Characteristic / Mechanism |
| --- | --- | --- |
| **Classic Gradient Descent** | **Batch Gradient Descent** | Computes the error gradient using the **entire dataset** per step. Extremely slow for large datasets. |
|  | **Stochastic Gradient Descent (SGD)** | Updates weights using **one sample** at a time. Fast, but highly noisy and erratic. |
|  | **Mini-batch Gradient Descent** | The standard version of SGD; uses small batches (e.g., 32, 64, 128 samples). Strikes a balance between speed and stability. |
| **Momentum-Based** | **Momentum** | Adds a fraction of the previous weight update to the current update. Helps the model build velocity down slopes and blast through local traps/plateaus. |
|  | **Nesterov Accelerated Gradient (NAG)** | An intelligent upgrade to Momentum. It "looks ahead" by calculating gradients at the anticipated future position, allowing it to brake earlier when approaching a minimum. |
| **Adaptive Learning Rate** | **AdaGrad (Adaptive Gradient)** | Adapts the learning rate individually for each parameter—giving smaller updates for frequent features and larger updates for infrequent ones. *Drawback:* Learning rate shrinks aggressively to zero over time. |
|  | **RMSprop** | Fixes AdaGrad's shrinking learning rate problem by using an exponentially decaying average of squared gradients instead of a cumulative sum. |
|  | **AdaDelta** | An extension of RMSprop that restricts the window of accumulated past gradients to a fixed size, preventing the learning rate from dropping to zero entirely. |
| **Hybrid (Momentum + Adaptive)** | **Adam (Adaptive Moment Estimation)** | Combines Momentum (first moment) and RMSprop (second moment). Highly reliable, fast-converging, and works out-of-the-box for most architectures. |
|  | **AdamW (Adam with Weight Decay)** | A corrected version of Adam that decouples weight decay (L2 regularization) from gradient updates, leading to significantly better generalization in deep neural networks (especially Transformers). |
|  | **Nadam (Nesterov-accelerated Adaptive Moment Estimation)** | Combines Adam with Nesterov momentum for faster convergence in the final stages of training. |
|  | **AMSGrad** | A variant of Adam that fixes a theoretical convergence issue in Adam by retaining past maximums of squared gradients. |
|  | **AdaMax** | A variant of Adam based on the infinity norm, offering stable updates for certain sparse data distributions. |
| **Second-Order Methods** | **L-BFGS (Limited-memory BFGS)** | Uses second-order derivatives (Hessian matrix approximation) to navigate curvature. Rarely used for massive deep networks due to high memory overhead, but powerful for small-to-medium datasets. |
