When evaluating why a machine learning model makes a specific prediction, explainability techniques operate across two fundamental scopes: **Global** (understanding overall model behavior across the entire dataset) and **Local** (explaining a single prediction for one specific instance).

---

### Method Comparison Matrix

| Feature / Metric | Standard Feature Importance | LIME (Local Linear Surrogate) | SHAP (Shapley Additive Explanations) |
| --- | --- | --- | --- |
| **Explanation Scope** | **Global** (Dataset-wide) | **Local** (Single instance) | **Both** (Local instances aggregate into global summaries) |
| **Core Mechanism** | Measures loss in performance when a feature is removed or permuted | Fits a weighted linear model around perturbed samples near $x$ | Calculates fair game-theoretic Shapley values across all feature subsets |
| **Model Agnostic?** | Model-dependent (MDI) or Model-Agnostic (Permutation) | Fully Model-Agnostic | Model-Agnostic (KernelSHAP) with tree-specific acceleration (TreeSHAP) |
| **Mathematical Guarantees** | None (biased toward high-cardinality features) | None (susceptible to sampling noise across runs) | **Guaranteed** (Satisfies Efficiency, Symmetry, Dummy, and Additivity axioms) |
| **Computational Speed** | Fast (calculated during or right after training) | Fast (requires sampling ~1,000 perturbed instances) | Fast for trees ($O(TLD^2)$ polynomial time); slow for complex model-agnostic kernels |

---

### Architectural Roles: When to Use Which

#### 1. Standard Feature Importance (Permutation / Impurity)

* **What it answers:** *"Which variables drive the overall model performance across all training samples?"*
* **Best used for:** Initial feature selection, identifying redundant variables, and macro-level sanity checking during model training.
* **Key limitation:** Global importance can mask local variations. A feature with low global importance might still be the decisive factor in an edge-case prediction for a specific user.

#### 2. LIME (Local Interpretable Model-agnostic Explanations)

* **What it answers:** *"How does the model behave in the immediate local decision boundary surrounding this single applicant?"*
* **Best used for:** Fast local approximations, explaining predictions from third-party or black-box API endpoints, and unstructured data (text token removal, image superpixels).
* **Key limitation:** Stochastic instability—because LIME samples randomly around the target point, slightly different feature weights can be produced across separate runs.

#### 3. SHAP (SHapley Additive exPlanations)

* **What it answers:** *"How many exact units or probability points did each feature contribute to moving the prediction away from the baseline average?"*
* **Best used for:** Highly regulated domains (credit scoring, clinical risk, insurance underwriting), adverse action compliance, feature interaction detection, and unified local-to-global audits.
* **Key limitation:** Exact Shapley calculation requires evaluating $2^{\vert{}F\vert{}}$ feature combinations, making KernelSHAP computationally expensive on very high-dimensional data without tree-specific optimizations.
