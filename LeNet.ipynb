**LeNet** (specifically **LeNet-5**) is one of the earliest and most influential Convolutional Neural Network. Developed by Yann LeCun and his colleagues in 1998, it laid the foundational architectural blueprint for modern computer vision and deep learning systems used today.

---

### Core Architecture & Structure

LeNet-5 is a relatively small network by today's standards—consisting of **7 layers** (excluding the input layer) that alternate between feature extraction and classification:

1. **Input Layer:** Takes a 32×32 pixel grayscale image (originally designed for recognizing handwritten digits like ZIP codes and checks).
2. **Convolutional Layer (C1):** Applies six 5×5 filters to extract low-level features like edges and corners, outputting six feature maps (28×28).
3. **Average Pooling Layer (S2):** Subsamples the feature maps (reducing spatial resolution by half to 14×14) to provide translation invariance.
4. **Convolutional Layer (C3):** A second convolutional layer with 16 filters to capture more complex, higher-level patterns.
5. **Average Pooling Layer (S4):** Further subsamples the feature maps down to 5×5.
6. **Fully Connected Layer (C5 / F6):** Flattened feature vectors connected to dense layers (containing 120 and then 84 neurons).
7. **Output Layer:** Uses a Softmax classifier to predict the digit class (0 through 9).

---

### Historical Significance & Innovations

Before LeNet, most neural networks struggled with images because treating every pixel as an independent input led to an unmanageable number of weights. LeNet introduced three crucial concepts that define modern CNNs:

* **Local Receptive Fields:** Neurons in early layers only look at small, localized patches of the input image rather than the whole picture.
* **Shared Weights:** Filters reuse the same weights across different parts of the image, drastically reducing the total number of parameters and enabling translation invariance.
* **Subsampling (Pooling):** Reducing spatial resolution step-by-step makes the network robust to slight shifts, scaling, and distortions in the image.

---

### Legacy

While LeNet-5 was successfully deployed by banks to read handwritten checks in the late 1990s, its growth was limited at the time by small datasets and a lack of powerful GPUs. However, its core structure directly inspired deeper, modern architectures like **AlexNet** (which won the ImageNet competition in 2012 and ignited the modern AI boom).

### LeNet Architecture Breakdown

LeNet-5 is structured as a sequential pipeline of alternating convolutional feature extraction and subsampling layers, ending in fully connected classification layers.

| Layer | Type | Map Size | Kernel Size / Stride | Activation / Notes |
| --- | --- | --- | --- | --- |
| **Input** | Image | 32 × 32 | — | Grayscale input (normalized pixel values) |
| **C1** | Convolution | 28 × 28 | 5 × 5 (6 filters) | Extracts local primitive features |
| **S2** | Subsampling | 14 × 14 | 2 × 2 (Stride 2) | Average pooling with trainable weights/bias |
| **C3** | Convolution | 10 × 10 | 5 × 5 (16 filters) | Complex feature combination across feature maps |
| **S4** | Subsampling | 5 × 5 | 2 × 2 (Stride 2) | Second average pooling stage |
| **C5** | Convolution | 1 × 1 | 5 × 5 (120 filters) | Acts as a fully connected layer due to exact dimensions match |
| **F6** | Dense (FC) | 1 × 84 | — | Fully connected layer with 84 units |
| **Output** | Output | 1 × 10 | — | Euclidean Radial Basis Function (RBF) network units |

---

### Key Architectural Characteristics

* **Asymmetric C3 Layer:** Unlike modern CNNs where every channel in a convolutional layer connects to all channels in the previous layer, LeNet-5's C3 layer used a sparse connection scheme. Certain feature maps in C3 were only connected to specific subsets of S2 maps to break symmetry and keep computation low.
* **Non-linear Activations:** It utilized specialized activation functions (such as scaled hyperbolic tangent variants) rather than modern standards like ReLU.
* **Parameter Sharing:** By forcing local receptive fields to share weights across the entire visual field, LeNet drastically slashed the parameter count compared to fully connected MLPs, making image processing computationally feasible in the 1990s.

---

### Evolution Over Time: From LeNet to Modern Deep Learning

The progression from LeNet-5 to modern architectures highlights how deep learning solved LeNet's limitations—specifically vanishing gradients, hardware constraints, and scaling bottlenecks.

| Architecture Era | Key Advancement Over LeNet | Representative Models | Primary Impact |
| --- | --- | --- | --- |
| **Classic Era (1998)** | Introduction of spatial weight sharing and pooling. | **LeNet-5** | Proved that neural networks can process raw pixel grids directly. |
| **Deep Revival (2012)** | ReLU activations, heavy GPU parallelization, and dropout. | **AlexNet** | Sparked the modern deep learning boom by dominating ImageNet. |
| **Varying Depth (2014)** | Ultra-small 3×3 convolutions stacked deep to capture fine details. | **VGGNet** | Showed that network depth is a primary driver of representation power. |
| **Residual Learning (2015)** | Skip connections (residual blocks) to bypass vanishing gradients. | **ResNet** | Enabled training of hundreds or thousands of layers without degradation. |
| **Attention & Transformers (2020s)** | Self-attention mechanisms replacing strict local convolutions. | **Vision Transformers (ViT)** | Treats image patches like words in a sentence, scaling globally. |

### 1. PyTorch Implementation: LeNet-5 vs. ResNet-18

Below are the PyTorch implementations for both architectures configured for a standard benchmark dataset like **MNIST** (32×32 grayscale images).

#### LeNet-5 Implementation

LeNet-5 uses classic average pooling, hyperbolic tangent ($\tanh$) activations, and fully connected dense layers at the output.

```python
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms


# 1. Define the LeNet-5 Architecture
class LeNet5(nn.Module):

  def __init__(self, num_classes=10):
    super(LeNet5, self).__init__()
    # Input shape: [Batch_Size, 1, 32, 32]
    self.features = nn.Sequential(
        nn.Conv2d(
            in_channels=1, out_channels=6, kernel_size=5, stride=1, padding=0
        ),
        nn.Tanh(),
        nn.AvgPool2d(kernel_size=2, stride=2),
        nn.Conv2d(in_channels=6, out_channels=16, kernel_size=5, stride=1),
        nn.Tanh(),
        nn.AvgPool2d(kernel_size=2, stride=2),
        nn.Conv2d(
            in_channels=16, out_channels=120, kernel_size=5
        ),  # Acts as FC layer
        nn.Tanh(),
    )

    self.classifier = nn.Sequential(
        nn.Linear(120, 84),
        nn.Tanh(),
        nn.Linear(84, num_classes),
    )

  def forward(self, x):
    x = self.features(x)
    x = torch.flatten(x, 1)  # Flatten to [Batch, 120]
    x = self.classifier(x)
    return x


# 2. Setup Device, Data, and Hyperparameters
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"[INFO] Using device: {device}")

# MNIST images are 28x28, resize to 32x32 to fit LeNet-5 input spec
transform = transforms.Compose([
    transforms.Resize((32, 32)),
    transforms.ToTensor(),
    transforms.Normalize((0.1307,), (0.3081,)),
])

train_dataset = datasets.MNIST(
    root="./data", train=True, download=True, transform=transform
)
test_dataset = datasets.MNIST(
    root="./data", train=False, download=True, transform=transform
)

train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False)

# Initialize model, loss function, and optimizer
model = LeNet5(num_classes=10).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# 3. Training Loop with Logging Metrics
num_epochs = 5
train_losses = []
train_accuracies = []

print("\n--- Starting Training Process ---")
for epoch in range(num_epochs):
  model.train()
  running_loss = 0.0
  correct_predictions = 0
  total_samples = 0

  for batch_idx, (data, targets) in enumerate(train_loader):
    data, targets = data.to(device), targets.to(device)

    # Forward pass
    optimizer.zero_grad()
    outputs = model(data)
    loss = criterion(outputs, targets)

    # Backward pass & optimization
    loss.backward()
    optimizer.step()

    # Track metrics
    running_loss += loss.item() * data.size(0)
    _, predicted = torch.max(outputs, 1)
    correct_predictions += (predicted == targets).sum().item()
    total_samples += targets.size(0)

    # Print batch-level log every 200 batches
    if (batch_idx + 1) % 200 == 0:
      print(
          f"Epoch [{epoch+1}/{num_epochs}] | Batch [{batch_idx+1}/{len(train_loader)}] | "
          f"Batch Loss: {loss.item():.4f}"
      )

  # Calculate epoch-level metrics
  epoch_loss = running_loss / total_samples
  epoch_acc = (correct_predictions / total_samples) * 100
  train_losses.append(epoch_loss)
  train_accuracies.append(epoch_acc)

  print(
      f"--> Epoch [{epoch+1}/{num_epochs}] Finished | "
      f"Average Loss: {epoch_loss:.4f} | Accuracy: {epoch_acc:.2f}%"
  )

print("\n--- Training Complete ---")

# 4. Generate and Display Graphs using Matplotlib
plt.figure(figsize=(12, 5))

# Plot Training Loss
plt.subplot(1, 2, 1)
plt.plot(
    range(1, num_epochs + 1),
    train_losses,
    marker="o",
    color="b",
    label="Train Loss",
)
plt.title("Training Loss per Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.legend()

# Plot Training Accuracy
plt.subplot(1, 2, 2)
plt.plot(
    range(1, num_epochs + 1),
    train_accuracies,
    marker="s",
    color="g",
    label="Train Accuracy",
)
plt.title("Training Accuracy per Epoch")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.savefig("lenet5_training_metrics.png")  # Saves the graph to disk
print("[INFO] Saved training curves plot as 'lenet5_training_metrics.png'")
plt.show()  # Displays the graph popup window if run in a GUI environment
```

#### ResNet-18 (Modern Counterpart)

Using `torchvision.models`, we can instantiate a ResNet-18 model and adapt its first convolutional layer and final fully connected layer to handle 32×32 grayscale input.

```python
import torchvision.models as models

def get_custom_resnet18(num_classes=10):
  # Load ResNet-18 without pre-trained weights
  model = models.resnet18(weights=None)

  # Adapt first conv layer for 1-channel (grayscale) and smaller 32x32 images (stride=1, smaller kernel)
  model.conv1 = nn.Conv2d(
      1, 64, kernel_size=3, stride=1, padding=1, bias=False
  )
  model.maxpool = nn.Identity()  # Remove maxpool to preserve spatial dimensions

  # Replace final fully connected layer
  model.fc = nn.Linear(model.fc.in_features, num_classes)
  return model
```

### 2. Training Pipeline

Both models can be trained using a standard PyTorch training loop with Cross-Entropy Loss and the Adam optimizer:

```python
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms

# 1. Data Preparation (MNIST)
transform = transforms.Compose([
    transforms.Resize((32, 32)),
    transforms.ToTensor(),
    transforms.Normalize((0.1307,), (0.3081,)),
])
train_dataset = datasets.MNIST(
    root="./data", train=True, download=True, transform=transform
)
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)

# 2. Model, Loss, Optimizer Setup
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = LeNet5().to(device)  # or get_custom_resnet18()
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# 3. Training Loop
model.train()
for epoch in range(3):  # Example 3 epochs
  for batch_idx, (data, targets) in enumerate(train_loader):
    data, targets = data.to(device), targets.to(device)

    optimizer.zero_grad()
    output = model(data)
    loss = criterion(output, targets)
    loss.backward()
    optimizer.step()

  print(f"Epoch {epoch+1} Complete. Loss: {loss.item():.4f}")
```

### 3. Performance & Architecture Comparison

| Feature | LeNet-5 (1998) | ResNet-18 (2015) |
| --- | --- | --- |
| **Depth** | 7 layers | 18 layers (deep residual blocks) |
| **Parameter Count** | ~60,000 parameters | ~11.2 million parameters |
| **Activation Functions** | $\tanh$ / Sigmoid | ReLU (Rectified Linear Unit) |
| **Pooling Type** | Average Pooling | Max Pooling & Global Average Pooling |
| **Handling Vanishing Gradients** | Relies on shallow depth and small size | **Skip Connections (Residual Blocks)** |
| **Typical MNIST Accuracy** | ~98.5%–99.0% | ~99.4%–99.7% |
| **Complex Dataset Capability** | Fails heavily on complex datasets (e.g., CIFAR-10, ImageNet) | Excels at complex, high-resolution datasets |

### 4. Advantages of Modern Architectures (Why ResNet Wins)

1. **Residual Skip Connections ($F(x) + x$):**
In deep networks, gradients vanish or explode as they backpropagate through many layers. ResNet introduces "shortcut connections" that bypass one or more layers, allowing gradients to flow directly back to earlier layers unimpeded.
2. **ReLU Activations over $\tanh$:**
$\tanh$ saturates at high and low inputs, flattening its derivative to near zero and stalling gradient updates. ReLU maintains a constant gradient of 1 for positive inputs, enabling much faster and deeper learning.
3. **Batch Normalization:**
Modern architectures incorporate normalization layers after convolutions to stabilize internal covariate shift, allowing higher learning rates and reducing dependency on hyperparameter initialization.
4. **Hierarchical Feature Representation:**
While LeNet extracts basic edges and blobs, ResNet builds deep hierarchies of abstraction—allowing it to understand complex textures, parts, and holistic semantic structures in images.
